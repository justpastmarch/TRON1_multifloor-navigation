# 01. 2D 지도 생성 및 저장

## 이 Notebook으로 하는 일

원격 TRON1 센서 스택을 확인하고, 노트북 PC에서 `slam_gmapping`을 실행해 `/map`을 만든 뒤 `.pgm`과 `.yaml`로 저장합니다. 원격 `sensor_integration/wf_mapping.launch`는 이름과 달리 SLAM이 아니라 카메라, MID360, TF, wheel odometry, LaserScan을 제공하며 실제 SLAM은 이 Notebook이 시작합니다.

## 셀 실행 방법

회색 code cell은 **Python 코드**입니다. 일반 terminal에 붙여 넣지 말고 Jupyter에서 `Shift+Enter`로 한 셀씩 실행하십시오. 각 셀은 앞 셀에서 만든 `CONFIG`, `ROS_ENV`, process 변수를 사용하므로 중간 셀부터 실행하지 않습니다. 오류가 나오면 다음 셀로 넘어가지 말고 외부 조건을 수정한 뒤 실패한 셀부터 다시 실행합니다.

## 시작 전 필수 조건

먼저 Jupyter를 시작한 terminal에서 `bash notebooks/check_environment.sh mapping`을 실행해 `[FAIL]`이 없는지 확인합니다. 이 Notebook은 sensor stack과 SLAM은 다루지만 로봇을 움직이는 teleop/controller는 제공하지 않습니다. **현장에서 승인된 별도 controller와 물리적 비상 정지가 준비되지 않았다면 여기서 중단합니다.**

1. 환경 설정 셀을 실행합니다.
2. 다음 작은 설정 셀에서 controller가 준비된 뒤에만 `MAPPING_CONTROLLER_READY = True`로 바꾸고 `MAP_NAME`을 실제 층·날짜 이름으로 지정합니다.
3. 센서 셀은 `/scan`이 있으면 재사용하고, 없으면 mini PC의 검증된 sensor launch를 SSH로 시작합니다. 파일을 설치하거나 복사하지는 않습니다.
4. 충돌 검사 후 SLAM을 시작합니다. Navigation node가 있으면 중단합니다.
5. 여기서 셀 실행을 멈추고 승인된 controller로 천천히 주행합니다.
6. map 상태를 확인하고 충분히 완성된 뒤 저장합니다.
7. 마지막 셀로 이 kernel이 시작한 SLAM만 종료합니다. 원격 sensor stack은 유지됩니다.

`Run All`은 SLAM 시작 직후 빈 지도를 저장할 수 있으므로 사용하지 마십시오.

```python
from pathlib import Path
import os
import signal
import subprocess
import sys
import time

WORKSPACE = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
CONFIG_PATH = WORKSPACE / "config.env"
ROS_SETUP = Path("/opt/ros/noetic/setup.bash")
DEVEL_SETUP = WORKSPACE / "devel/setup.bash"

if not CONFIG_PATH.is_file() or not ROS_SETUP.is_file() or not DEVEL_SETUP.is_file():
    raise RuntimeError("workspace, config.env, or ROS Noetic setup is missing")

def read_config(path):
    values = {}
    for raw_line in path.read_text(encoding="utf-8").splitlines():
        line = raw_line.strip()
        if line and not line.startswith("#"):
            key, value = line.split("=", 1)
            values[key] = value
    return values

def sourced_environment():
    command = f"source {ROS_SETUP} && source {DEVEL_SETUP} && env -0"
    completed = subprocess.run(["bash", "-lc", command], check=True, stdout=subprocess.PIPE)
    environment = os.environ.copy()
    for entry in completed.stdout.split(b"\0"):
        if entry and b"=" in entry:
            key, value = entry.split(b"=", 1)
            environment[key.decode()] = value.decode()
    return environment

CONFIG = read_config(CONFIG_PATH)
ROS_ENV = sourced_environment()
ROS_ENV["ROS_MASTER_URI"] = f"http://{CONFIG['ROS_MASTER_HOST']}:{CONFIG['ROS_MASTER_PORT']}"
route = subprocess.run(["ip", "-4", "route", "get", CONFIG["ROS_MASTER_HOST"]], check=True, text=True, stdout=subprocess.PIPE).stdout.split()
ROS_ENV["ROS_IP"] = route[route.index("src") + 1]
ROS_ENV.pop("ROS_HOSTNAME", None)

for path in (Path("/opt/ros/noetic/lib/python3/dist-packages"), WORKSPACE / "devel/lib/python3/dist-packages"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

MAP_OUTPUT_DIR = WORKSPACE / "stair_captures/maps"
MAP_NAME = "floor_new"  # 예: floor_4F_20260827
MAPPING_PROCESS = None
print(f"workspace={WORKSPACE}")
print(f"ROS_MASTER_URI={ROS_ENV['ROS_MASTER_URI']} ROS_IP={ROS_ENV['ROS_IP']}")
```

```python
MAP_NAME = "floor_new"  # 예: floor_4F_20260827
MAPPING_CONTROLLER_READY = False  # 승인된 controller와 물리적 비상 정지 준비 후 True
print(f"map output={MAP_OUTPUT_DIR / MAP_NAME}")
print(f"mapping controller ready={MAPPING_CONTROLLER_READY}")
```

## 센서 스택 재사용 또는 시작

이 셀은 **Python**에서 terminal command를 실행합니다. `/scan`이 있으면 원격 프로세스를 건드리지 않습니다. 없을 때만 `config.env`의 SSH 주소로 접속해 mini PC의 기존 `wf_mapping.launch`를 background로 시작합니다. package 설치나 파일 복사는 하지 않습니다. 성공하면 `/scan/header`가 출력되며, 실패하면 mini PC의 `/tmp/tron1-notebook-sensors.log`를 확인하고 이 셀부터 다시 실행합니다.

```python
def ros_output(argv, timeout=10):
    return subprocess.run(argv, env=ROS_ENV, check=True, text=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE, timeout=timeout).stdout.strip()

def topic_type(topic):
    completed = subprocess.run(["rostopic", "type", topic], env=ROS_ENV, text=True, stdout=subprocess.PIPE, stderr=subprocess.DEVNULL, timeout=3)
    return completed.stdout.strip() if completed.returncode == 0 else ""

def wait_for_topic(topic, expected_type, timeout=30):
    deadline = time.monotonic() + timeout
    while time.monotonic() < deadline:
        if topic_type(topic) == expected_type:
            return
        time.sleep(0.5)
    raise RuntimeError(f"topic unavailable or wrong type: {topic} expected={expected_type}")

if not topic_type("/scan"):
    target = f"{CONFIG['MINI_PC_USER']}@{CONFIG['MINI_PC_HOST']}"
    remote = (
        f"source /opt/ros/noetic/setup.bash; source {CONFIG['MINI_PC_WORKSPACE']}/devel/setup.bash; "
        f"export ROS_MASTER_URI={ROS_ENV['ROS_MASTER_URI']}; export ROS_IP={CONFIG['MINI_PC_ROS_IP']}; "
        f"export D435F_SERIAL={CONFIG['D435F_SERIAL']}; unset ROS_HOSTNAME; "
        "pgrep -f '[r]oslaunch.*wf_mapping.launch' >/dev/null || "
        "nohup setsid roslaunch sensor_integration wf_mapping.launch >/tmp/tron1-notebook-sensors.log 2>&1 </dev/null &"
    )
    subprocess.run(["ssh", "-o", "BatchMode=yes", "-o", "ConnectTimeout=5", target, remote], check=True, timeout=15)

wait_for_topic("/scan", "sensor_msgs/LaserScan")
wait_for_topic("/tron/wheel_odom_raw", "nav_msgs/Odometry")
print(ros_output(["rostopic", "echo", "-n", "1", "/scan/header"], timeout=10))
```

## 충돌 검사와 SLAM 시작

정적 지도를 제공하는 `/map_server`, AMCL, move_base 또는 다른 SLAM이 하나라도 실행 중이면 mapping을 시작하지 않습니다. 소유 terminal이 있으면 그 terminal에서 `Ctrl+C`로 종료합니다. 이전에 Kernel이 비정상 종료되어 이 Notebook의 `/slam_gmapping`만 남았다고 확인한 경우 `rosnode kill /slam_gmapping`으로 정리한 뒤 이 셀을 다시 실행하십시오. 소유권을 모르는 node는 종료하지 않습니다.

```python
if not MAPPING_CONTROLLER_READY:
    raise RuntimeError("mapping controller is not armed; prepare the approved controller and E-stop, then set MAPPING_CONTROLLER_READY = True")
nodes = set(ros_output(["rosnode", "list"]).splitlines())
conflicts = {"/map_server", "/amcl", "/move_base", "/slam_gmapping", "/hector_mapping", "/slam_toolbox"} & nodes
if conflicts:
    raise RuntimeError(f"mapping과 충돌하는 node를 먼저 종료하십시오: {sorted(conflicts)}")

scan_info = ros_output(["rostopic", "info", "/scan"])
publishers_section = scan_info.split("Publishers:", 1)[1].split("Subscribers:", 1)[0]
publisher_lines = [line for line in publishers_section.splitlines() if line.startswith(" * /")]
if len(publisher_lines) != 1:
    raise RuntimeError(f"/scan publisher는 정확히 하나여야 합니다: {publisher_lines}")

tf_check = subprocess.run(["timeout", "2s", "rosrun", "tf", "tf_echo", "odom", "base_Link"], env=ROS_ENV, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
if tf_check.returncode not in {0, 124} or "Translation:" not in tf_check.stdout:
    raise RuntimeError(f"odom -> base_Link TF is unavailable: {tf_check.stderr.strip()}")
MAPPING_PROCESS = subprocess.Popen(
    [
        "rosrun", "gmapping", "slam_gmapping",
        "scan:=/scan", "_base_frame:=base_Link", "_odom_frame:=odom",
        "_map_frame:=map", "_map_update_interval:=2.0",
    ],
    env=ROS_ENV, stdout=subprocess.DEVNULL, stderr=subprocess.STDOUT, start_new_session=True,
)
wait_for_topic("/map", "nav_msgs/OccupancyGrid", timeout=30)
print(f"slam_gmapping started: pid={MAPPING_PROCESS.pid}")
print(ros_output(["rostopic", "echo", "-n", "1", "/map/info"], timeout=10))
print("NEXT: stop running cells, drive slowly with the approved controller, then run the map status cell")
```

## 지도 작성

이제 로봇을 천천히 이동시키며 복도와 방의 벽을 충분히 관측합니다. 급회전, 유리·거울 근처의 장시간 정지, 같은 구간의 고속 반복 주행은 map distortion을 만들 수 있습니다. `/map`의 width/height가 증가하고 loop가 자연스럽게 닫히는지 확인한 다음 저장 셀을 실행하십시오.

```python
print(ros_output(["rostopic", "echo", "-n", "1", "/map/info"], timeout=10))
print(ros_output(["rostopic", "echo", "-n", "1", "/map/header"], timeout=10))
```

## 지도 저장과 다음 단계

지도 coverage를 눈으로 확인한 뒤 실행합니다. `MAP_NAME`은 층과 날짜를 알 수 있는 새 이름이어야 하며 결과는 `stair_captures/maps/<MAP_NAME>.pgm`과 `.yaml`입니다. 기존 파일은 덮어쓰지 않습니다.

성공 기준은 두 파일이 모두 생성되고 마지막 출력에 `saved:`가 표시되는 것입니다. 이 파일은 바로 운영 지도가 아닙니다. 현장 검증 후에만 두 파일을 `src/multifloor_manager/config/maps/`로 복사하고 해당 floor의 `floors.yaml`을 갱신한 뒤 workspace root에서 `python3 validate_bundle.py --normalize-map`과 production validation을 실행합니다.

```python
if not MAP_NAME or MAP_NAME in {"floor_new", "map"}:
    raise RuntimeError("MAP_NAME을 실제 층/날짜가 포함된 이름으로 변경하십시오")
MAP_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
prefix = MAP_OUTPUT_DIR / MAP_NAME
outputs = [prefix.with_suffix(".pgm"), prefix.with_suffix(".yaml")]
if any(path.exists() for path in outputs):
    raise RuntimeError(f"existing map would be overwritten: {outputs}")
subprocess.run(["rosrun", "map_server", "map_saver", "-f", str(prefix)], env=ROS_ENV, check=True, timeout=30)
for path in outputs:
    if not path.is_file() or path.stat().st_size == 0:
        raise RuntimeError(f"map_saver did not create a valid file: {path}")
print("saved:", *outputs, sep="\n- ")
```

## 종료

이 셀은 현재 Notebook이 시작한 `slam_gmapping` process group에만 SIGINT를 보냅니다. 원격 센서 스택은 다음 작업에서 재사용할 수 있도록 유지합니다.

```python
if MAPPING_PROCESS is not None and MAPPING_PROCESS.poll() is None:
    os.killpg(MAPPING_PROCESS.pid, signal.SIGINT)
    MAPPING_PROCESS.wait(timeout=10)
    print("slam_gmapping stopped")
else:
    print("this kernel owns no running slam_gmapping process")
```
